```python
parser=StrOutputParser()
def llm_qa(state :LMMstate) ->LMMstate:
    question =state['question']
    prompt =PromptTemplate(template="can you write essay in 50 words about this topic\n {question}",
                           input_variables='question')
    #prompt_invoke=prompt.invoke({"question": "question"})
    chain= prompt |model |parser 
    state['answer']=chain.invoke ({"question": question})
    #state['answer']=model.invoke(prompt_invoke).content
    return state
```

```python
graph=StateGraph(LMMstate)
graph.add_node('LLM',llm_qa)
graph.add_edge(START,'LLM')
graph.add_edge('LLM',END)
```

```python
workflow=graph.compile()
initial_state ={'question':'cricket'}
final_state=workflow.invoke(initial_state)
```
